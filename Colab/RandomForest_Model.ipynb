```python
import numpy as np
from pathlib import Path

features_path = Path(
    "/content/drive/MyDrive/Paddy_Disease_Project/features/_MobileNetV2"
)

X_train = np.load(features_path / "X_train.npy")
y_train = np.load(features_path / "y_train.npy")
X_valid = np.load(features_path / "X_valid.npy")
y_valid = np.load(features_path / "y_valid.npy")

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_valid:", X_valid.shape)
print("y_valid:", y_valid.shape)
```

```python
classes = [
    'bacterial_leaf_blight',
    'bacterial_leaf_streak',
    'bacterial_panicle_blight',
    'blast',
    'brown_spot',
    'dead_heart',
    'downy_mildew',
    'hispa',
    'normal',
    'tungro'
]
```

```python


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import time

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    class_weight='balanced'
)

# Training
start_time = time.time()
rf_model.fit(X_train, y_train)
rf_train_time = time.time() - start_time

# Prediction
start_time = time.time()
rf_preds = rf_model.predict(X_valid)
rf_inference_time = time.time() - start_time

# Evaluation
rf_accuracy = accuracy_score(y_valid, rf_preds)

print(f"Random Forest Training time: {rf_train_time:.4f} seconds")
print(f"Random Forest Inference time: {rf_inference_time:.4f} seconds")
print(f"\nRandom Forest Accuracy: {rf_accuracy * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(
    y_valid,
    rf_preds,
    target_names=classes,
    digits=4
))
```

```python
import joblib

model_path = "/content/drive/MyDrive/Paddy_Disease_Project/models"

joblib.dump(
    rf_model,
    f"{model_path}/model_4_random_forest.pkl"
)

print("Random Forest model saved successfully.")
```
